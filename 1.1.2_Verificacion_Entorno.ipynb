###**Paso 1: Preparación del Entorno e Importación de Librerías**

```python
import numpy as np
import pandas as pd
import os
```

###**Paso 2: Descarga de Datos de Kaggle y Carga en Memoria**

```python
# Carga inicial controlando delimitadores y codificación estándar
df_clientes = pd.read_csv('ecommerce_customer_data.csv', sep=',',
encoding='utf-8')
# Inspección de las dimensiones de la matriz de datos
print(f"Dimensiones del dataset cargado: {df_clientes.shape}")
```

###**Paso 3: Perfilamiento Estructural del Dataset**

```python
df_clientes.info()
```

```python
df_clientes.describe()
```

###**Paso 4: Computación Vectorial con NumPy**

```python
#Extraer columna EstimatedSalary como un arreglo Numpy
montos_brutos = df_clientes['EstimatedSalary'].to_numpy()

#Calcular gasto neto de operación con tasa impositiva fija del 19% + un costo operativo logistico.
gasto_neto = montos_brutos * 1.19 + 2.5

#Guardamos el resultado en una nueva columna del DataFrame
df_clientes['GastoNetoOperación'] = gasto_neto
```

###**Paso 5: Segmentación y Aislamiento Avanzado**

```python
#Extraer segmento clientes + su gasto neto
clientes_gastos = df_clientes.iloc[49:100, [1, 14]]

print(clientes_gastos)
```

###**Paso 6: Análisis de Eficiencia Física**

**Exportar a Excel**

```python
#Guardamos el dataset actualizado
df_clientes.to_csv(
    'ecommerce_clientes_resultado.csv',
    index=False,
    encoding='utf-8'
)

print("Archivo CSV creado correctamente.")
```
